## Erdős-Rényi

```python
# ── Parametros ──────────────────────────────────────────────
NUM_NODES = 500
EDGE_PROB = 0.02
SEED      = 42
# ────────────────────────────────────────────────────────────

g = ErdosRenyiGraph(num_nodes=NUM_NODES, edge_prob=EDGE_PROB, seed=SEED)
show_full(g.graph, f'Erdos-Renyi  G({NUM_NODES}, p={EDGE_PROB})', seed=SEED)
```

## Watts-Strogatz (Small World)

```python
# ── Parametros ──────────────────────────────────────────────
NUM_NODES   = 500
K           = 6
REWIRE_PROB = 0.1
SEED        = 42
# ────────────────────────────────────────────────────────────

g = WattsStrogatzGraph(num_nodes=NUM_NODES, k=K, rewire_prob=REWIRE_PROB, seed=SEED)
show_full(g.graph, f'Watts-Strogatz  n={NUM_NODES}  k={K}  p={REWIRE_PROB}', seed=SEED)
```

## Scale-Free (Barabási-Albert extendido)

```python
# ── Parametros ──────────────────────────────────────────────
NUM_NODES  = 500
ALPHA      = 0.41
BETA       = 0.54
GAMMA      = 0.05
DELTA_IN   = 0.2
DELTA_OUT  = 0.0
SEED       = 42
# ────────────────────────────────────────────────────────────

g = ScaleFreeGraph(
    num_nodes=NUM_NODES, alpha=ALPHA, beta=BETA, gamma=GAMMA,
    delta_in=DELTA_IN, delta_out=DELTA_OUT, seed=SEED,
)
show_full(
    g.graph,
    f'Scale-Free  n={NUM_NODES}  a={ALPHA}  b={BETA}  g={GAMMA}',
    seed=SEED,
)
```

## Multilayer — ScaleFree (digital) + Watts-Strogatz (analogica)

```python
# ── Parametros ──────────────────────────────────────────────
NUM_NODES    = 300
# Capa digital (ScaleFree)
SF_ALPHA     = 0.41
SF_BETA      = 0.54
SF_GAMMA     = 0.05
SF_DELTA_IN  = 0.2
SF_DELTA_OUT = 0.0
# Capa analogica (Watts-Strogatz)
WS_K         = 6
WS_REWIRE    = 0.1
SEED         = 42
# ────────────────────────────────────────────────────────────

digital = ScaleFreeGraph(
    num_nodes=NUM_NODES, alpha=SF_ALPHA, beta=SF_BETA, gamma=SF_GAMMA,
    delta_in=SF_DELTA_IN, delta_out=SF_DELTA_OUT, seed=SEED,
)
analog = WattsStrogatzGraph(num_nodes=NUM_NODES, k=WS_K, rewire_prob=WS_REWIRE, seed=SEED)
ml = MultiLayerGraph(digital_graph=digital, analog_graph=analog, seed=SEED)

show_multilayer(ml.graph, f'Multilayer ScaleFree+WS  n={NUM_NODES}', seed=SEED)
```

## Multilayer — SNAP (digital) + Watts-Strogatz (analogica)

```python
# ── Parametros ──────────────────────────────────────────────
DATASET   = 'ego-Facebook'     # ~4 039 nodos; se descarga si no esta en cache
CACHE_DIR = '../../data/snap'
# Capa analogica (Watts-Strogatz) — num_nodes se fija al tamanyo real del dataset
WS_K      = 6
WS_REWIRE = 0.1
SEED      = 42
# ────────────────────────────────────────────────────────────

# Capa digital: SNAP siempre dirigido en multilayer
digital = SNAPGraph(dataset_name=DATASET, cache_dir=CACHE_DIR, directed=True, seed=SEED)
n = len(digital)   # dispara la descarga/parseo; obtiene el nº de nodos real

# Capa analogica con exactamente los mismos n nodos
analog = WattsStrogatzGraph(num_nodes=n, k=WS_K, rewire_prob=WS_REWIRE, seed=SEED)
ml = MultiLayerGraph(digital_graph=digital, analog_graph=analog, seed=SEED)

show_multilayer(ml.graph, f'Multilayer SNAP({DATASET})+WS  n={n}', seed=SEED)
```
